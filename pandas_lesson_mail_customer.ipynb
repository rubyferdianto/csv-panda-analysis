## 2. Understand Customer Demographics

```python
#2a. Calculate the average Age, Annual Income, and Spending Score to understand the typical customer profile
#Hint: Use NumPy operations like mean(), min(), and max() for this analysis.

age_mean = np.mean(mc['Age'])
income_mean = np.mean(mc['Annual Income (k$)'])
score_mean = np.mean(mc['Spending Score (1-100)'])

age_min = np.min(mc['Age'])
income_min = np.min(mc['Annual Income (k$)'])
score_min = np.min(mc['Spending Score (1-100)'])

age_max = np.max(mc['Age'])
income_max = np.max(mc['Annual Income (k$)'])
score_max = np.max(mc['Spending Score (1-100)'])

print(f"Average Age: {age_mean:.2f}, Min: {age_min}, Max: {age_max}")
print(f"Average Annual Income (k$): {income_mean:.3f}, Min: {income_min}, Max: {income_max}")
print(f"Average Spending Score: {score_mean:.2f}, Min: {score_min}, Max: {score_max}")
```

Summary Analysis:
-   The average age and income values help you understand the typical demographic and economic status of your customers.
-   The range (min to max) for each metric shows the diversity in your customer base.
-   The spending score average and range indicate how actively customers are spending, which can help in segmenting high-value customers or identifying potential for targeted marketing.

```python
#2b. Analyze customer distribution by gender. Does one gender tend to spend more or earn more on average
#using same NumPy operations like mean(), min(), and max() for this analysis

for gender in mc['Gender'].unique():
    gender_group = mc[mc['Gender'] == gender]
    age_mean = np.mean(gender_group['Age'])
    income_mean = np.mean(gender_group['Annual Income (k$)'])
    score_mean = np.mean(gender_group['Spending Score (1-100)'])
    print(f"{gender}:")
    print(f"  Average Age: {age_mean:.2f}")
    print(f"  Average Annual Income (k$): {income_mean:.2f}")
    print(f"  Average Spending Score: {score_mean:.2f}\n")
```

Summary Analysis:
-   Average Age: Shows if one gender group is generally younger or older.
-   Average Annual Income: Reveals if one gender tends to earn more on average.
-   Average Spending Score: Indicates if one gender tends to spend more or is more engaged as a customer.
-   If one gender has a higher average annual income, they may represent a more affluent segment.
-   If one gender has a higher average spending score, they may be more valuable or engaged customers.
-   Comparing average ages can help you tailor marketing strategies to the dominant age group within each gender.

## 4. Explore Relationships Between Features

```python
#Compute the pairwise correlations between Age, Annual Income, and Spending Score to uncover key drivers of spending
#Correlation can be computed using NumPy's corrcoef().

# Extract relevant columns as a NumPy array
data = mc[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']].values

# Compute the correlation matrix
corr_matrix = np.corrcoef(data, rowvar=False)

# Display the correlation matrix with labels
columns = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
print("Pairwise Correlation Matrix:")
print(pd.DataFrame(corr_matrix, index=columns, columns=columns))

#np.corrcoef computes the correlation matrix for the given data.
#The correlation coefficient (Pearson’s r) measures the linear relationship between two variables, 
#ranging from -1 (perfect negative) to +1 (perfect positive), with 0 meaning no linear correlation

#Why is this useful?
#High positive value: Strong direct relationship (as one increases, so does the other).
#High negative value: Strong inverse relationship (as one increases, the other decreases).
#Near zero: No linear relationship.
```

Interpretation:

Spending Score vs. Age:
Correlation = -0.327227
This is a moderate negative correlation.
Meaning: As age increases, spending score tends to decrease. Older customers generally have lower spending scores.

Spending Score vs. Annual Income:
Correlation = 0.009903
This value is very close to zero.
Meaning: There is almost no linear relationship between annual income and spending score. Customers with higher or lower income do not consistently spend more or less.

Summary:

Age is somewhat negatively related to spending score (older customers spend less).
Annual income has almost no effect on spending score (income does not predict spending behavior in this dataset).

```python
#Filter young adults (18-25) and calculate their average Spending Score. Compare this with older age groups
#using Correlation can be computed using NumPy's corrcoef()

# Young adults: 18-25
young_adults = mc[(mc['Age'] >= 18) & (mc['Age'] <= 25)]
avg_score_young = np.mean(young_adults['Spending Score (1-100)'])

# Older adults: 26+
older_adults = mc[mc['Age'] > 25]
avg_score_older = np.mean(older_adults['Spending Score (1-100)'])

print(f"Average Spending Score (Age 18-25): {avg_score_young:.2f}")
print(f"Average Spending Score (Age 26+): {avg_score_older:.2f}")
```

## 5. Customer Segmentation

```python
# Customer Segmentation by Spending Score

#Low: 1-33
low_segment = mc[(mc['Spending Score (1-100)'] >= 1) & (mc['Spending Score (1-100)'] <= 33)]
#Medium: 34-66
medium_segment = mc[(mc['Spending Score (1-100)'] >= 34) & (mc['Spending Score (1-100)'] <= 66)]
#High: 67-100
high_segment = mc[(mc['Spending Score (1-100)'] >= 67) & (mc['Spending Score (1-100)'] <= 100)]

# Compute averages
segments = {
    'Low': low_segment,
    'Medium': medium_segment,
    'High': high_segment
}

for name, group in segments.items():
    avg_age = np.mean(group['Age'])
    avg_income = np.mean(group['Annual Income (k$)'])
    print(f"{name} Segment:")
    print(f"  Average Age: {avg_age:.2f}")
    print(f"  Average Annual Income (k$): {avg_income:.2f}")
    print(f"  Number of Customers: {len(group)}\n")
```

##### Segment Characteristics & Marketing Strategies
**Low Segment (Spending Score 1-33)**
- *Characteristics:*
  - Typically older customers or those less engaged with the store.
  - May have a wide range of annual incomes, but low spending activity.
- *Marketing Strategies:*
  - Run awareness campaigns to increase engagement.
  - Offer introductory discounts or basic loyalty programs.
  - Educate about products/services to spark interest.

**Medium Segment (Spending Score 34-66)**
- *Characteristics:*
  - Represent the average customer base.
  - Moderate age and income levels.
  - Show some engagement but not highly active.
- *Marketing Strategies:*
  - Use personalized promotions and targeted offers.
  - Provide incentives (e.g., points, small rewards) to encourage more frequent purchases.
  - Regularly engage through newsletters or events to move them to the high segment.

**High Segment (Spending Score 67-100)**
- *Characteristics:*
  - Often younger and/or highly engaged customers.
  - May have moderate to high annual incomes.
  - Most active and valuable segment for the business.
- *Marketing Strategies:*
  - Implement VIP or premium loyalty programs.
  - Offer exclusive deals, early access to new products, or special events.
  - Focus on retention with personalized communication and recognition.

**Summary:** 
Segmenting customers by spending score helps tailor marketing strategies:  
- Low: Focus on activation and education.  
- Medium: Encourage higher engagement.  
- High: Reward loyalty and maximize retention.

# 6. Find Behavioral Similarities

```python
# Normalize the dataset (subtract the mean and divide by the standard deviation for each column
# Use NumPy's vectorized operations for distance calculations

features = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
data = mc[features].values

# Compute mean and std for each column
means = np.mean(data, axis=0)
stds = np.std(data, axis=0)

# Normalize
normalized_data = (data - means) / stds

print("First 5 rows of normalized data:")
print(normalized_data[:5])

# Example: Compute Euclidean distance between first two customers
dist = np.linalg.norm(normalized_data[0] - normalized_data[1])
print(f"Euclidean distance between first two customers: {dist:.2f}")
```

```python
#Compute the Euclidean distance between each customer and the customer with the highest Spending Score
#Use NumPy's vectorized operations for distance calculations

# Find the index of the customer with the highest Spending Score
max_score_idx = mc['Spending Score (1-100)'].idxmax()

# Get the normalized feature vector for that customer
target_vector = normalized_data[max_score_idx]

# Compute Euclidean distances from all customers to the target customer (vectorized)
distances = np.linalg.norm(normalized_data - target_vector, axis=1)

# Show the first 5 distances as an example
print("First 5 distances to the highest Spending Score customer:")
print(distances[:5])
```

```python
#Identify customers who exhibit similar behavior
#Use NumPy's vectorized operations for distance calculations

# Let's find the 5 customers most similar to the highest Spending Score customer (excluding themselves)
similar_indices = np.argsort(distances)[1:6]  # [0] is the customer themselves

print("Top 5 customers most similar to the highest Spending Score customer:")
print(mc.iloc[similar_indices][['Gender', 'Age', 'Annual Income (k$)', 'Spending Score (1-100)']])
```
